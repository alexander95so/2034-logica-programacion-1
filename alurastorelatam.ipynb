### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda_2 = pd.read_csv(url2)
tienda_3 = pd.read_csv(url3)
tienda_4 = pd.read_csv(url4)

tienda.head()

```

# 4. Productos más y menos vendidos

###Tienda 1

```python

productos = tienda['Producto'].value_counts()
productos.plot(kind='barh', figsize=(10,10), color='green', xlabel='Cantidad', ylabel='Producto')
```

```python


max_tienda = productos.max()
min_tienda = productos.min()
destacados = productos[(productos == max_tienda) | (productos == min_tienda)]



destacados.plot(kind='barh', figsize=(5,5), color='green', xlabel='Cantidad', ylabel='Producto')
```

###Tienda 2

```python

productos2 = tienda_2['Producto'].value_counts()
productos2.plot(kind='barh', figsize=(10,10), color='black', xlabel='Cantidad', ylabel='Producto')
```

```python

max_tienda_2 = productos2.max()
min_tienda_2 = productos2.min()
destacados2 = productos2[(productos2 == max_tienda_2) | (productos2 == min_tienda_2)]
destacados2.plot(kind='barh', figsize=(3,3), color='brown', xlabel='Cantidad', ylabel='Producto')
```

###Tienda 3

```python

productos3 = tienda_3['Producto'].value_counts()
productos3.plot(kind='barh', figsize=(10,10), color='blue', xlabel='Cantidad', ylabel='Producto')
```
